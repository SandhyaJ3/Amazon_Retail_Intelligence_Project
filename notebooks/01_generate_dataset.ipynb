```python
import pandas as pd
import numpy as np
import random
from datetime import datetime, timedelta
```

```python
random.seed(42)
np.random.seed(42)
```

```python
categories = ["Electronics","Fashion","Books","Home","Grocery"]
payment_methods = ["UPI","Credit Card","Debit Card","Cash on Delivery","Net Banking"]
regions = ["North","South","East","West"]
```

```python
orders = []
```

```python
start_date = datetime(2025,1,1)
end_date = datetime(2026,12,31)
```

```python
for order_id in range(1,101): #generates number 1 to 100
   quantity = random.randint(1,5)
   price = random.randint(200,5000)
   sales = quantity * price
   random_days = random.randint(0,(end_date-start_date).days)
   order_date = start_date + timedelta(days=random_days)
   order = {"Order_ID": order_id,
            "Order_Date": order_date,
            "Category": random.choice(categories),
            "Region": random.choice(regions),
            "Payment_Method": random.choice(payment_methods),
            "Quantity":quantity,
            "Unit_Price":price,
            "Sales":sales
   }
   orders.append(order)  #Take the current order dictionary , add to list
   
```

```python
orders
```

```python
df = pd.DataFrame(orders)
df.head()
```
