```python
import pandas as pd
import numpy as np
import random
from datetime import datetime, timedelta
```

```python
random.seed(42)
np.random.seed(42)
```

```python
categories = ["Electronics","Fashion","Books","Home","Grocery"]
payment_methods = ["UPI","Credit Card","Debit Card","Cash on Delivery","Net Banking"]
regions = ["North","South","East","West"]
```

```python
orders = []
```

```python
start_date = datetime(2025,1,1)
end_date = datetime(2026,12,31)
```

```python
for order_id in range(1,101): #generates number 1 to 100
   quantity = random.randint(1,5)
   price = random.randint(200,5000)
   sales = quantity * price
   random_days = random.randint(0,(end_date-start_date).days)
   order_date = start_date + timedelta(days=random_days)
   order = {"Order_ID": order_id,
            "Order_Date": order_date,
            #"Category": random.choice(categories),
            #"Region": random.choice(regions),
            "Payment_Method": random.choice(payment_methods),
            "Quantity":quantity
   }
   orders.append(order)  #Take the current order dictionary , add to list
   
```

Generate tables for Customers & Product

```python
num_customers = 100
customers = pd.DataFrame({
    "customer_id":range(1,num_customers+1),
    "customer_name":[f"Customer_{i}" for i in range(1, num_customers + 1)],
    "city":np.random.choice(["Bangalore","Delhi","Chennai","Mumbai"],num_customers),
    "signup_date":pd.to_datetime("2024-01-01")  + pd.to_timedelta(np.random.randint(0,365,num_customers),unit="D")
})
```

```python
num_products = 50
products = pd.DataFrame({
    "product_id":range(1,num_products+1),
    "product_name":[f"Product_{i}" for i in range(1,num_products+1)],
    "category": np.random.choice(["Electronics","Clothing","Home","Books"],num_products),
    "price": np.random.randint(100,5000,num_products)
})
```

```python
df["customer_id"] = np.random.randint(1, num_customers + 1, len(df))
df["product_id"] = np.random.randint(1, num_products + 1, len(df))
df.head()
```

```python
df.to_csv("../data/raw/amazon_retail_data.csv",index=False)
```

```python
customers.to_csv("../data/raw/customers.csv", index=False)
```

```python
products.to_csv("../data/raw/products.csv", index=False)
```

```python
df = df.merge(customers, on="customer_id", how="left")
df = df.merge(products, on="product_id", how="left")
```

```python
df["Sales"] = df["Quantity"] * df["price"]
df.head()
```
